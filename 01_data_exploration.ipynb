## 4. Data Quality Check

```python
# Check for missing values
print('Missing Values:')
missing = df.isnull().sum()
missing[missing > 0]
```

```python
# Visualize missing data
plt.figure(figsize=(12, 6))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
plt.title('Missing Data Heatmap')
plt.tight_layout()
plt.show()
```

```python
# Check for duplicates
duplicates = df.duplicated().sum()
print(f'\nNumber of duplicate rows: {duplicates}')
```

## 5. User and Language Distribution

```python
# Number of unique users
if 'user_id' in df.columns:
    n_users = df['user_id'].nunique()
    print(f'Total unique users: {n_users}')
    
    # Samples per user
    samples_per_user = df['user_id'].value_counts()
    print(f'\nSamples per user - Mean: {samples_per_user.mean():.1f}, Median: {samples_per_user.median():.1f}')
    print(f'Min: {samples_per_user.min()}, Max: {samples_per_user.max()}')
```

```python
# Language distribution
if 'language' in df.columns:
    print('\nLanguage Distribution:')
    lang_dist = df['language'].value_counts()
    print(lang_dist)
    
    # Visualize
    plt.figure(figsize=(10, 6))
    lang_dist.plot(kind='bar', color='steelblue')
    plt.title('Distribution of Samples by Language', fontsize=14)
    plt.xlabel('Language')
    plt.ylabel('Number of Samples')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

## 6. Keystroke Timing Analysis

```python
# Dwell time distribution
if 'dwell_time' in df.columns:
    plt.figure(figsize=(14, 5))
    
    # Histogram
    plt.subplot(1, 2, 1)
    plt.hist(df['dwell_time'].dropna(), bins=50, color='skyblue', edgecolor='black')
    plt.title('Dwell Time Distribution', fontsize=14)
    plt.xlabel('Dwell Time (ms)')
    plt.ylabel('Frequency')
    
    # Box plot
    plt.subplot(1, 2, 2)
    plt.boxplot(df['dwell_time'].dropna())
    plt.title('Dwell Time Box Plot', fontsize=14)
    plt.ylabel('Dwell Time (ms)')
    
    plt.tight_layout()
    plt.show()
    
    print(f'\nDwell Time Statistics:')
    print(df['dwell_time'].describe())
```

```python
# Flight time distribution
if 'flight_time' in df.columns:
    plt.figure(figsize=(14, 5))
    
    # Histogram
    plt.subplot(1, 2, 1)
    plt.hist(df['flight_time'].dropna(), bins=50, color='lightcoral', edgecolor='black')
    plt.title('Flight Time Distribution', fontsize=14)
    plt.xlabel('Flight Time (ms)')
    plt.ylabel('Frequency')
    
    # Box plot
    plt.subplot(1, 2, 2)
    plt.boxplot(df['flight_time'].dropna())
    plt.title('Flight Time Box Plot', fontsize=14)
    plt.ylabel('Flight Time (ms)')
    
    plt.tight_layout()
    plt.show()
    
    print(f'\nFlight Time Statistics:')
    print(df['flight_time'].describe())
```

## 7. Cross-Language Comparison

```python
# Compare dwell time across languages
if 'language' in df.columns and 'dwell_time' in df.columns:
    plt.figure(figsize=(12, 6))
    
    # Box plot by language
    df.boxplot(column='dwell_time', by='language', figsize=(10, 6))
    plt.title('Dwell Time Comparison Across Languages')
    plt.suptitle('')  # Remove default title
    plt.ylabel('Dwell Time (ms)')
    plt.xlabel('Language')
    plt.tight_layout()
    plt.show()
    
    # Statistical comparison
    print('\nDwell Time by Language:')
    print(df.groupby('language')['dwell_time'].describe())
```

```python
# Compare flight time across languages
if 'language' in df.columns and 'flight_time' in df.columns:
    plt.figure(figsize=(12, 6))
    
    df.boxplot(column='flight_time', by='language', figsize=(10, 6))
    plt.title('Flight Time Comparison Across Languages')
    plt.suptitle('')
    plt.ylabel('Flight Time (ms)')
    plt.xlabel('Language')
    plt.tight_layout()
    plt.show()
    
    print('\nFlight Time by Language:')
    print(df.groupby('language')['flight_time'].describe())
```

## 8. Correlation Analysis

```python
# Select numeric columns for correlation
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()

if len(numeric_cols) > 1:
    # Compute correlation matrix
    corr_matrix = df[numeric_cols].corr()
    
    # Visualize
    plt.figure(figsize=(12, 10))
    sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, 
                square=True, linewidths=1, cbar_kws={"shrink": 0.8})
    plt.title('Feature Correlation Matrix', fontsize=16)
    plt.tight_layout()
    plt.show()
```

## 9. Sample User Analysis

```python
# Analyze one user's typing pattern
if 'user_id' in df.columns:
    sample_user = df['user_id'].iloc[0]
    user_data = df[df['user_id'] == sample_user]
    
    print(f'\nSample User: {sample_user}')
    print(f'Total samples: {len(user_data)}')
    
    if 'language' in user_data.columns:
        print(f'Languages: {user_data["language"].unique()}')
    
    # Plot typing pattern over time
    if 'dwell_time' in user_data.columns and 'timestamp' in user_data.columns:
        user_data_sorted = user_data.sort_values('timestamp')
        
        plt.figure(figsize=(14, 5))
        plt.plot(range(len(user_data_sorted)), user_data_sorted['dwell_time'], marker='o', markersize=3)
        plt.title(f'Typing Pattern for User {sample_user}', fontsize=14)
        plt.xlabel('Keystroke Number')
        plt.ylabel('Dwell Time (ms)')
        plt.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()
```
